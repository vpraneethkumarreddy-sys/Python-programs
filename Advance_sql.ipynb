```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('company.db')
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS Employees (
    EmpID INTEGER PRIMARY KEY,
    EmpName TEXT,
    Department TEXT,
    Age INTEGER,
    Salary REAL
)
""")

cursor.executemany("""
INSERT INTO Employees VALUES (?, ?, ?, ?, ?)
""", [
    (1, 'Ravi', 'IT', 25, 50000),
    (2, 'Sita', 'HR', 30, 45000),
    (3, 'Kiran', 'IT', 28, 60000),
    (4, 'Anu', 'Finance', 35, 70000)
])

conn.commit()

df = pd.read_sql_query("SELECT * FROM Employees", conn)
print(df)

conn.close()
```

```python
#Group By

import sqlite3
import pandas as pd

conn = sqlite3.connect('company.db')

query = """
SELECT Department, SUM(Salary) AS TotalSalary
FROM Employees
GROUP BY Department
"""

df = pd.read_sql_query(query, conn)
print(df)

conn.close()
```

```python
#Having
import sqlite3
import pandas as pd

conn = sqlite3.connect('company.db')

query = """
SELECT Department, AVG(Salary) AS AvgSalary
FROM Employees
GROUP BY Department
HAVING AVG(Salary) > 50000
"""

df = pd.read_sql_query(query, conn)
print(df)

conn.close()
```
