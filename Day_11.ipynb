```python
#merge
import pandas as pd
data1 = [[1, 'Raj', 25],

         [2, 'Sharma', 42],

         [3, 'Saroj', 22],

         [4, 'Raja', 51],

         [5, 'Kajal', 26]]

df1 = pd.DataFrame(data1, columns=['ID', 'Name', 'Age'])

data2 = [[1, 'Male', 25000],

         [2, 'Male', 42500],

         [3, 'Female', 22300],

         [4, 'Male', 51400],

         [5, 'Female', 26800]]

df2 = pd.DataFrame(data2, columns=['ID', 'Gender', 'Salary'])
merged_df = pd.merge(df1, df2, on='ID', how='inner')

print(merged_df)
```

```python
#right merge
import pandas as pd
data1 = [[1, 'Raj', 25],

         [2, 'Sharma', 42],

         [3, 'Saroj', 22],

         [4, 'Raja', 51],

         [5, 'Kajal', 26]]

df1 = pd.DataFrame(data1, columns=['ID', 'Name', 'Age'])

data2 = [[1, 'Male', 25000],

         [2, 'Male', 42500],

         [3, 'Female', 22300],

         [4, 'Male', 51400],

         [5, 'Female', 26800]]

df2 = pd.DataFrame(data2, columns=['ID', 'Gender', 'Salary'])
merged_df = pd.merge(df1, df2, on='ID', how='right')

print(merged_df)
```

```python
#outer merge
import pandas as pd
data1 = [[1, 'Raj', 25],

         [2, 'Sharma', 42],

         [3, 'Saroj', 22],

         [4, 'Raja', 51],

         [5, 'Kajal', 26]]

df1 = pd.DataFrame(data1, columns=['ID', 'Name', 'Age'])

data2 = [[1, 'Male', 25000],

         [2, 'Male', 42500],

         [3, 'Female', 22300],

         [4, 'Male', 51400],

         [5, 'Female', 26800]]

df2 = pd.DataFrame(data2, columns=['ID', 'Gender', 'Salary'])
merged_df = pd.merge(df1, df2, on='ID', how='outer')

print(merged_df)
```

```python
#pivot
import pandas as pd
df = pd.DataFrame({
    'A': ['John', 'Boby', 'Mina'],
    'B': ['Masters', 'Graduate', 'Graduate'],
    'C': [27, 23, 21]
})

pivot_df = df.pivot(index='A', columns='B', values='C')

print(pivot_df)
```

```python
#left merge
import pandas as pd
data1 = [[1, 'Raj', 25],

         [2, 'Sharma', 42],

         [3, 'Saroj', 22],

         [4, 'Raja', 51],

         [5, 'Kajal', 26]]

df1 = pd.DataFrame(data1, columns=['ID', 'Name', 'Age'])

data2 = [[1, 'Male', 25000],

         [2, 'Male', 42500],

         [3, 'Female', 22300],

         [4, 'Male', 51400],

         [5, 'Female', 26800]]

df2 = pd.DataFrame(data2, columns=['ID', 'Gender', 'Salary'])
merged_df = pd.merge(df1, df2, on='ID', how='left')

print(merged_df)
```
